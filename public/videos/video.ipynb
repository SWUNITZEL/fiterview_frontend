```python
import cv2

# 비디오 파일 열기
cap = cv2.VideoCapture("interviewer_1.mp4")

# 프레임 총 개수 가져오기
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# 프레임 저장 리스트
frames = []

# 모든 프레임 읽어서 리스트에 저장
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

cap.release()

# 코덱 및 출력 설정
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
height, width, _ = frames[0].shape
out = cv2.VideoWriter("output_reverse.mp4", fourcc, 30, (width, height))

# 프레임을 역순으로 저장
for frame in reversed(frames):
    out.write(frame)

out.release()
print("역재생 비디오 저장 완료!")
```

```python
import requests
from tqdm import tqdm

univ_data = []
for i in tqdm(range(1,30)):
    # API 요청 URL
    url = "https://openapi.gg.go.kr/Univmjrm"
    params = {
        "KEY": "d2879d07f08c481fa4b0e5ccf03208f2",    
        "Type": "json",
        "pSize": 1000,
        "pIndex": i
    }

    # 데이터 요청
    response = requests.get(url, params=params)
    data = response.json()

    items = data['Univmjrm'][1]['row']  # 실제 데이터는 두 번째 요소의 'row' 리스트 안에 있음
    # SCHOOL_DIV_NM이 '대학교'인 항목만 필터링
    univ_data.extend([item for item in items 
                      if (item['SCHOOL_DIV_NM'] == '대학교')
                      &(item['EXAMIN_YY'] == '2020')
                      &(item['DAN_DIV_NM'] == '주간')
                      &(item['MJR_STATE_NM'] != '폐과')
                      &(item['DEG_CRSE_NM'] == '학사')
                      ])
```

```python
# 2️⃣ 대학별로 학과 모으기
univ_dict = {}
for row in univ_data:
    school = row.get('SCHOOL_NM')
    dept = row.get('UNDSTUD_KWA_NM')
    if not school or not dept:
        continue
    univ_dict.setdefault(school, set()).add(dept)

# 3️⃣ 원하는 형태로 변환
result = [
    {
        "id": idx + 2,
        "name": school,
        "departments": sorted(list(depts))
    }
    for idx, (school, depts) in enumerate(univ_dict.items())
]
```
